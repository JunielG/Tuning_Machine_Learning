```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn import linear_model
```

```python
df = pd.read_csv('canada_per_capita_income.csv')
df.head()
```

```python
df_copy = df.copy()
```

```python
# Changing column name 
df = df.rename(columns={'income_per_capita (US)': 'income_per_capita'})
```

```python
# Rounding to two decimals number
df['income_per_capita'] = round(df['income_per_capita'], 2)
```

```python
df.head()
```

```python
df.isna().sum()
```

```python
df.info()
```

```python
%matplotlib inline
plt.xlabel('year')
plt.ylabel('income_per_capita (US)')
plt.scatter(df.year, df.income_per_capita, color='red',marker='+')
plt.show()
```

```python
X = df[['year']]
y = df['income_per_capita (US)']
```

```python
linear_reg = linear_model.LinearRegression()
```

```python
linear_reg.fit(X, y)
```

```python
linear_reg.predict([[2020]])
```

```python
linear_reg.coef_
```

```python
linear_reg.intercept_
```

```python
%matplotlib inline
plt.xlabel('year')
plt.ylabel('income_per_capita (US)')
plt.scatter(df.year, df.income_per_capita, color='red',marker='+')
plt.plot(df.year, linear_reg.predict(df[['year']]), color='blue')
plt.show()
```
