```python
df = pd.read_csv('home_prices.csv')
```

```python
df
```

```python
sns.scatterplot(x=df['area'], y=df['price'], color='red')
plt.xlabel('area(sqr ft)')
plt.ylabel('price(US$)')
plt.show()
```

```python
lg = linear_model.LinearRegression()
```

```python
lg.fit(df[['area']], df.price)
```

```python
lg.predict([[3500]])
```

```python
lg.coef_
```

```python
lg.intercept_
```

```python
# This is the operation used to get 
135.78767123 * 3500 + 180616.43835616432
```

```python
areas = pd.read_csv('areas.csv')
```

```python
prices_pred = lg.predict(areas)
```

```python
areas['prices'] = prices_pred
```
