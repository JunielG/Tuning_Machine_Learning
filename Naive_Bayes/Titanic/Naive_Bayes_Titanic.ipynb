```python
%matplotlib inline
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt
from sklearn.naive_bayes import GaussianNB
from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv('titanic.csv')
```

```python
df.head()
```

```python
pass_id = df['PassengerId']
```

```python
df = df.drop(columns=['PassengerId', 'Name', 'SibSp', 'Parch', 'Ticket', 'Cabin', 'Embarked'])
```

```python
df.head()
```

```python
df.isna().sum()
```

```python
mean_age = df['Age'].mean()
median_age = df['Age'].median()
mode_age = df['Age'].mode()
```

```python
print(mean_age, '\n', median_age, '\n', mode_age)
```

```python
df = round(df.fillna({'Age': mean_age}), 2)
```

```python
df.isna().sum()
```

```python
df = pd.get_dummies(df, dtype='int')
```

```python
df.head()
```

```python
df = df.rename(columns={'Sex_female': 'female',	'Sex_male': 'male'})
```

```python
df.head()
```

```python
X = df.drop('Survived', axis=1)
y = df['Survived']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size= 0.3)
```

```python
nb_model = GaussianNB()
nb_model.fit(X_train, y_train)
nb_model.score(X_test, y_test)
```

```python
pred = nb_model.predict(X_test)
pred
```

```python
nb_model.predict_proba(X_test[:10])
```

```python
from sklearn.model_selection import cross_val_score
cross_val_score(GaussianNB(),X_train, y_train, cv=5)
```

```python
# Create a DataFrame with the features from X_test
final_df = pd.DataFrame(X_test, columns=['Pclass', 'Age', 'Fare', 'female', 'male'])
```

```python
# Add the predictions as a new column
final_df['predicted'] = pred

# Add a new column
final_df['Passenger_Id'] = pass_id
```
