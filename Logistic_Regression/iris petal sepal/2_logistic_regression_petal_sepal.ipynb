Use sklearn.datasets iris flower dataset to train your model using logistic regression. You need to figure out accuracy of your model and use that to predict different samples in your test dataset. In iris dataset there are 150 samples containing following features,

Sepal Length

Sepal Width

Petal Length

Petal Width

Using above 4 features you will clasify a flower in one of the three categories,

Setosa

Versicolour

Virginica

```python
%matplotlib inline
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression, LogisticRegression, LogisticRegressionCV
from sklearn.preprocessing import LabelEncoder, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix
```

```python
from sklearn import datasets

iris = datasets.load_iris()
```

```python
_, ax = plt.subplots()
scatter = ax.scatter(iris.data[:, 0], iris.data[:, 1], c=iris.target)
ax.set(xlabel=iris.feature_names[0], ylabel=iris.feature_names[1])
_ = ax.legend(
    scatter.legend_elements()[0], iris.target_names, loc="lower right", title="Classes"
)
```

```python
data = iris.data
```

```python
target = iris.target
```

```python
target_names = iris.target_names
```

```python
X_train, X_test, y_train, y_test = train_test_split(data, target, test_size=0.2)
lg_reg = LogisticRegression()
lg_reg.fit(X_train, y_train)
```

```python
X_test
```

```python
pred = lg_reg.predict(X_test)
```

```python
pred
```

```python
lg_reg.score(X_test, y_test)
```

```python
X_test.shape
```

```python
pred.shape
```

```python
# Assuming X_test is your feature matrix with shape (30, 4)
# And pred is your predictions array with shape (30,)

# Create a DataFrame with the features from X_test
final_df = pd.DataFrame(X_test, columns=['feature1', 'feature2', 'feature3', 'feature4'])

# Add the predictions as a new column
final_df['predicted'] = pred
```

```python
plt.figure(figsize=(15, 10))
for i in range(4):  # Assuming X_test has 4 features
    plt.subplot(2, 2, i+1)  # Create a 2x2 grid of subplots
    plt.scatter(pred, X_test[:, i])
    plt.xlabel('Predicted Values')
    plt.ylabel(f'Feature {i+1} Values')
    plt.title(f'Predictions vs Feature {i+1}')
plt.tight_layout()
plt.show()
```
