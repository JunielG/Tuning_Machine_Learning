```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import LabelEncoder, OneHotEncoder
from sklearn.compose import ColumnTransformer
```

```python
df = pd.read_csv("carprices.csv")
```

```python
df = df.rename(columns={'Car Model': 'car_model', 'Mileage': 'mileage',
                   'Sell Price($)': 'sell_price', 'Age(yrs)': 'years'})
```

```python
df
```

```python
df = pd.get_dummies(df, dtype='int')
```

```python
df = df.drop(columns=['car_model_Mercedez Benz C class'], axis=1)
```

```python
df
```

```python
X = df.drop('sell_price', axis=1) 
y = df['sell_price']
```

```python
lg_model = LinearRegression()
lg_model.fit(X, y)
```

```python
lg_model.predict(X)
```

```python
lg_model.score(X, y)
```

```python
lg_model.predict([[45000, 4, 0, 0]]) # Mercedes price
```

```python
lg_model.predict([[45000, 2, 1, 0]]) # Audi price
```

```python
lg_model.predict([[86000, 7, 0, 1]]) # BMW price 
```
