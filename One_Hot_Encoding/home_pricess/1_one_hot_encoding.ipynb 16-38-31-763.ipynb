```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import LabelEncoder, OneHotEncoder
from sklearn.compose import ColumnTransformer
```

```python
df = pd.read_csv('homeprices.csv')
```

```python
df
```

```python
dummies = pd.get_dummies(df, dtype='int')
```

```python
dummies.head()
```

```python
final = dummies.drop('town_west windsor', axis=1)
```

```python
final
```

```python
X = final.drop('price', axis='columns')
y = final['price']
```

```python
lg_model = LinearRegression()
lg_model.fit(X, y)
```

```python
pred = lg_model.predict(X)
```

```python
lg_model.score(X, y)
```

```python
lg_model.predict([[2800, 1, 0]]) # 2800 sqr ft home in town_monroe township
```

```python
lg_model.predict([[2800, 0, 1]]) # 2800 sqr ft home in town_robinsville
```

```python
lg_model.predict([[3400, 0, 0]]) # 3400 sqr ft home in west windsor
```

```python
dataf = pd.DataFrame({'actual_price': y, 'predicted_price': pred})
```

## Label/One hot encoder 

```python
df1 = pd.read_csv('homeprices.csv')
```

```python
df1
```

```python
df1['town'] = np.where(df1['town'] == 'monroe township', 0, df1['town'])
df1['town'] = np.where(df1['town'] == 'west windsor', 2, df1['town'])
df1['town'] = np.where(df1['town'] == 'robinsville', 1, df1['town'])
```

```python
df1
```

```python
le = LabelEncoder()
```

```python
dfle = df1
dfle.town = le.fit_transform(dfle.town)
```

```python
dfle
```

```python
X = dfle[['town', 'area']].values
y = dfle['price'].values
```

```python
col_transf = ColumnTransformer([('town', OneHotEncoder(), [0])], remainder= 'passthrough')
```

```python
X = col_transf.fit_transform(X)
```

```python
X = X[:, 1:]
```

```python
lg_model.fit(X, y)
```

```python
lg_model.predict([[0, 1, 3400]]) # 3400 sqr ft home in monroe township / robinsville
```

```python
lg_model.predict([[1, 0, 2800]]) # 2800 sqr ft home in robbinsville / monroe township
```

```python
lg_model.predict([[2, 0, 4000]]) # 2800 sqr ft home in west windsor
```
